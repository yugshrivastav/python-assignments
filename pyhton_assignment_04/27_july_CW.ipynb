```python
from PIL import Image
import matplotlib.pyplot as plt

# Load Image
img = Image.open("nature.jpg")

# Display Original Image
plt.imshow(img)
plt.title("Original Nature Image")
plt.axis("off")
plt.show()

# Resize Image
resize_img = img.resize((400, 300))

plt.imshow(resize_img)
plt.title("Resized Nature Image")
plt.axis("off")
plt.show()

# Flip Image
flip_img = img.transpose(Image.FLIP_LEFT_RIGHT)

plt.imshow(flip_img)
plt.title("Flipped Nature Image")
plt.axis("off")
plt.show()
```

```python
import pandas as pd

data = {
    'Movie': [
        'Dangal',
        'Jawan',
        'Pathaan',
        '3 Idiots',
        'Bajrangi Bhaijaan'
    ],
    'Release_Year': [2016, 2023, 2023, 2009, 2015],
    'Box_Office_Crore': [2024, 1148, 1050, 460, 969],
    'IMDb_Rating': [8.3, 7.0, 6.0, 8.4, 8.1]
}

df = pd.DataFrame(data)

df
```

```python
df.info()

df.describe()

print("Highest Collection:", df["Box_Office_Crore"].max())
print("Lowest Collection:", df["Box_Office_Crore"].min())
print("Average Rating:", df["IMDb_Rating"].mean())
print("Median Release Year:", df["Release_Year"].median())
```

```python
sorted_movies = df.sort_values(by="Box_Office_Crore", ascending=False)
sorted_movies
```

```python
top_movies = df[df["IMDb_Rating"] > 8.0]
top_movies
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.bar(df["Movie"], df["Box_Office_Crore"])
plt.title("Box Office Collection")
plt.xlabel("Movie")
plt.ylabel("Collection (Crore)")
plt.xticks(rotation=20)
plt.show()
```

```python
plt.figure(figsize=(6,6))
plt.pie(
    df["Box_Office_Crore"],
    labels=df["Movie"],
    autopct="%1.1f%%"
)
plt.title("Box Office Share")
plt.show()
```
