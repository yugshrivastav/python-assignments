```python
numbers = [12, 5, 8, 21, 16, 9]

print("Even Numbers:")
for num in numbers:
    if num % 2 == 0:
        print(num)
```

```python
text = "I like Python"

print("Original:", text)

# Reverse the string
print("Reverse String:", text[::-1])

# Reverse the words
words = text.split()
words.reverse()
print("Reverse Words:", " ".join(words))
```
