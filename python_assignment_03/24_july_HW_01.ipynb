```python
import pandas as pd

matches = pd.read_csv("matches.csv")
deliveries = pd.read_csv("deliveries.csv")

print(matches.head())
print(deliveries.head())
```

```python
df = matches
print("\ndataset info:")
print(df.info())
print("\ndatatset statistics:")
print(df.describe())
```

```python
import matplotlib.pyplot as plt

# Toss winner count
matches['toss_winner'].value_counts().head(10).plot(kind='bar')

plt.title("Top 10 Teams Winning Toss")
plt.xlabel("Team")
plt.ylabel("Number of Toss Wins")
plt.xticks(rotation=90)
plt.grid(axis='y')
plt.show()
```

```python
import matplotlib.pyplot as plt

# Bar chart of the number of matches won by each team

team_wins = matches['winner'].value_counts()

plt.figure(figsize=(10,5))
plt.bar(team_wins.index, team_wins.values, color='gold')

plt.title("Number of Matches Won by Each Team")
plt.xlabel("Teams")
plt.ylabel("Number of Matches Won")

plt.xticks(rotation=90)
plt.grid(axis='y')

plt.show()
```

```python
import matplotlib.pyplot as plt

team_wins = matches['winner'].value_counts().head(5)

plt.figure(figsize=(7,7))

plt.pie(
    team_wins.values,
    labels=team_wins.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Top 5 Teams by Matches Won")
plt.axis('equal')

plt.show()
```
